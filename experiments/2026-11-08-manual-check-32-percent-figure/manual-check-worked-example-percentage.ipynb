```python
import math

def wilson(k, n, z=1.96):
    """95% Wilson score interval for a proportion k/n."""
    p = k / n
    centre = (p + z * z / (2 * n)) / (1 + z * z / n)
    half = z * math.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / (1 + z * z / n)
    return centre - half, centre + half

assert None not in marks, "some chunks are still unmarked"
k, n = sum(marks), len(marks)
lo, hi = wilson(k, n)
print(f"your audit: {k}/{n} = {k / n:.1%}  (95% CI {lo:.0%}–{hi:.0%})")

lo13, hi13 = wilson(13, 40)
print(f"June figure, 13/40: {13 / 40:.1%}  (95% CI {lo13:.0%}–{hi13:.0%})")
```
